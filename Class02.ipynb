# Supervised Learning
  - Simple Linear Regression(1 input,1 output)

```python
import pandas as pd
```

### Loading Dataset 

```python
data = pd.read_csv("weight-height.csv")
```

```python
data.head(5)
```

### Ploting Data

```python
data.plot(kind="scatter",x="Height",y='Weight',
         title="height and Weight in Adults")
```

### Spliting dataset into X,y

```python
data
```

```python
X = data[['Height']]
y=data[['Weight']]
```

### Splitting data into Training and Testing

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=32)
```

```python
X_train
```

```python
X_train.shape
```

```python
X_test.shape
```

### Model Selection 

```python
from sklearn.linear_model import LinearRegression
weight_predictor = LinearRegression()
```

###  Training the Model

```python
weight_predictor.fit(X_train,y_train)
```

```python
#slop value from model
weight_predictor.coef_ 
```

```python
#intercept value from model
weight_predictor.intercept_
```

# height=7.7(Weight)+(-349.8086)

```python
y_pred_train = weight_predictor.predict(X_train)
```

```python
import matplotlib.pyplot as plt
```

```python
data.plot(kind="scatter",x="Height",y="Weight",color="y")
plt.plot(X_train,y_pred_train,color="r")
plt.show()
```

# checking Error in our model

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
import math
math.sqrt(mean_squared_error(y_train,y_pred_train))
```

```python
r2_score(y_train,y_pred_train)
```

```python
y_pred_test = weight_predictor.predict(X_test)
```

```python
mean_squared_error(y_test,y_pred_test)
```

```python
r2_score(y_test,y_pred_test)
```

# Prediction 

```python
def weight_pred(height):
    return weight_predictor.predict([[height]])*0.453592
```

```python
height_inches = float(input("Enter Your Height "))*12
weight_pred(height_inches)
```
